```python
#Dictionary(Maps)

#its a collection of {key:values} pairs ordered and changeable. No duplicates
capitals = {"USA": "Washington D.C.",
                    "India": "New Delhi",
                    "China": "Beijing",
                    "Russia": "Moscow"}

# print(dir(capitals))
# print(help(capitals))
# print(capitals.get("Japan"))
# print(capitals["Russia"])

# if capitals.get("Japan"):
#     print(capitals["found"])
# else:
#     print("Not Found")

# capitals.update({"Germany": "Berlin"})
# capitals.update({"USA": "Las Vegas"})
# capitals.pop("China")
# capitals.popitem() #removes the latest updated values
# capitals.clear()
# print(capitals)

keys=capitals.keys()
for key in keys:
    print(key)

values=capitals.values()
for value in values:
    print(value)

items=capitals.items()
for item in items:
    print(item)
```

```python
#create a multi dimensional dictionary
mystock = {
    "product":"Earphones",
    "price": 800,
    "Quantity":50,
    "Instock":"yes"
}

mystock2 = {
    "product":"HDD",
    "price": 2500,
    "Quantity":50,
    "Instock":"yes"
}

print(mystock)
print(mystock2)

products = {1:mystock, 2:mystock2}
print(products)
```
