```python
#2D lists
#list.tuples : inner rows with () works
groceries = [("apple", "orange","banana","coconut"),("tomato", "potato", "cabbage"),("chicken", "beef", "pork")]

for collection in groceries:
    for item in collection:
        print(item, end=" ")
    print()
```

```python
#num_pad
num_pad = ((1,2,3),
          (4,5,6),
          (7,8,9),
          ("*", 0, "#"))

for row in num_pad:
    for num in row:
        print(num, end=" ")
    print()

```
