```python
#Python Shopping cart

foods = []
prices = []
total = 0

while True:
  food = input("Enter a food to buy (q to end): ")
  if food.lower() == "q":
    break
  else:
    price = float(input(f"Enter the price: {food}: $"))
    foods.append(food)
    prices.append(price)
print("-----SHOPPING CART-----")
for food in foods:
  print(food, end=" ")
for price in prices:
  total +=price
print()
print(f"Total: ${total}")
```
